#### Write a Python program to load an image file, display it using Matplotlib, and create a color-inverted version of the image. Display both the original and inverted images for comparison.

```
import numpy as np
import matplotlib.pyplot as plt
```

```
try:
    # logo = np.load('numpy-logo.png')
    logo = plt.imread('numpy-logo.png')
    plt.figure(figsize=(8, 4))
    plt.subplot(1, 2, 1)
    plt.grid(False)
    plt.title('Numpy Logo')
    plt.imshow(logo)

    # Dark logo
    dark_logo = logo.copy()
    dark_logo[..., :3] = 1 - dark_logo[..., :3]
    plt.subplot(1, 2, 2)
    plt.grid(False)
    plt.title('Numpy Dark Logo')
    plt.imshow(dark_logo)

except FileNotFoundError:
    print("Numpy Logo not found")


```
