### Creating array from list

```python
arr_1d = np.array([1, 2, 3, 4, 5])
print(f"1D array: {arr_1d}")

arr_2d = np.array([[10, 20, 30], [100, 200, 300]])
print(f"2D array: {arr_2d}")
```

```python
py_list = [1, 2, 3]
print("Python list multiplication")
print(py_list * 2)

print()

py_arr = np.array([1, 2, 3]) # element wise multiplication
print("Python array multiplication")
print(py_arr * 2)
```

```python
import time
import numpy as np

start = time.time()

my_list = [i * i for i in range(1000000)]
print(f"List operation time {time.time() - start}")

start = time.time()
my_arr = np.arange(1000000)
print(f"NumPy Array operation time {time.time() - start}")
```

### All Zero Element array
Inside zeros method we can use list [], or tuple ()

```python
zeros_arr = np.zeros([2, 3])
print(f"Zeros Array: \n {zeros_arr}")
```

### All Ones Element array
Inside zeros method we can use list [], or tuple ()

```python
ones_arr = np.ones((3, 3))
print(f"Ones Array: \n {ones_arr}")
```

### Full Array with all 10
Inside zeros method we can use list [], or tuple ()

```python
full_arr = np.full((2, 4), 10)
print(f"Full Array with all 10 elements: \n {full_arr}")
```
