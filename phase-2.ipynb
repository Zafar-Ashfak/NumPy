## Operations on Array

```
import numpy as np
```

### Slicing and Sorting On 1D Array or Vector

```
arr = np.array([1, 5, 2, 3, 5, 4, 6, 1, 9, 7, 4, 8])
print(arr)
print(arr[2 : 8]) # Index 8 is not inclusive
print(arr[0: ])
print(arr[: 6]) # Index 6 is not inclusive
print(arr[-9: -3]) # Negative indexing
print(arr[::-1]) # Reverse an array

# Sorting
print(f"Sorted Array: {np.sort(arr)}")
```

### Slicing and Sorting On 2D array or Matrix

```
arr_2D = np.array([[1, 2, 6, 2],
                   [4, 3, 1, 2],
                   [5, 1, 2, 3],
                   [2, 3, 1, 9]])
print(f"Specific Element: {arr_2D[0, 1]}")
print(f"Specific Element: {arr_2D[2, 3]}")
print(f"Entire row: {arr_2D[1]}") # Prints entire row
print(f"Entire Column: {arr_2D[:, 2]}") # Prints entire column

# Sorting
print(f"Sorted 2D Array by row: {np.sort(arr_2D, axis=1)}") # Row wise sorting
print(f"Sorted 2D Array by column: {np.sort(arr_2D, axis=0)}") # Column wise sorting
```

## Filtering in numpy array

```
nums = np.array([1, 4, 9, 3, 5, 2, 8, 1, 7, 3, 4, 8, 6, 9, 5])

even_nums = nums[nums % 2 == 0]
print(f"Even numbers are in nums: {even_nums}")
```

### Fiter with mask

```
mask = nums > 4
print(f"Numbers greater than 4: {nums[mask]}")
```

### Indices and Where()

```
indices = [1, 5, 7, 9, 13]
print(f"Numbers are at \n{indices} : \n{nums[indices]}")

# Where
where_res = np.where(nums % 2 != 0)
# print(f"Odd numbers in the array nums: {where_res}") # It prints with array itself
print(f"Odd numbers in the array nums: {nums[where_res]}") # It prints with array itself
```

### where(condition, x, y)

```
my_nums = np.array([2, 1, 5, 6, 3, 8, 4, 9, 3, 7, 1])

res = np.where(my_nums > 4, my_nums ** 2, my_nums ** 3)
print(res)
```

### Arrays Concatenation

```
arr1 = np.array([1, 3, 5, 7])
arr2 = np.array([2, 4, 6, 8])

# print(arr1 + arr2) # This statement adds the elements of the arrays

print(np.concatenate((arr1, arr2))) # This statement concat the arrays
```

### Compatibility Check

```
a = np.array([1, 2, 3])
b = np.array([4, 3, 9])
c = np.array([2, 1, 8, 7])

print(f"Compatibility: {a.shape == b.shape}") # True
print(f"Compatibility: {b.shape == c.shape}") # False
```
