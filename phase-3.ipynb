
# Advance operation with Business examples

```
import numpy as np
import matplotlib.pyplot as ptl
```

```
# Data Structure = [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 200000, 140000, 160000, 190000], # Beijing Bites
    [3, 120000, 210000, 260000, 300000], # Pizza Hub
    [4, 180000, 230000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 205000, 300000] # Chai Point
])

print("==== Zomato Sales Analysis ====")
print(f"Sales data shape: {sales_data.shape}")
```

```
print(f"Sales data for 1st 3 restaurant: {sales_data[ : 3]}")
```

### Can remove resturant_id with slicing

```
print(f"Restaurant data without id: {sales_data[:, 1:]}")
```

### Total Sales per year

```
print(f"Total sales per year: {np.sum(sales_data, axis=0)}") # calc sum column wise

yearly_sales = np.sum(sales_data[:, 1:], axis=0) # calc sum of sales col wise without restaurant id

print(yearly_sales)
```

### Calculate minimum sales per restaurant

```
min_sale = np.min(sales_data[:, 1:], axis=1)
print(f"Minimum sales per restaurant: {min_sale}")
```

### Calculate maximum sales per year

```
max_sales = np.max(sales_data[:, 1:], axis=0)
print(f"Maximum sales per year: {max_sales}")
```

### Calculate average sales per restaurant

```
avg_sales = np.average(sales_data[:, 1:], axis=1)
print(f"Average sales per restaurant: {avg_sales}")

mean_sales = np.mean(sales_data[:, 1:], axis=1) # mean() also calc average
print(f"Average sales per restaurant: {mean_sales}")
```

### Cumulative Sum
  => A cumulative sum is a running total of numbers in a sequence or data set.

  Imagine a list of daily sales: [10, 20, 5]
  Day 1: 10 (the total is just 10)

  Day 2: 10 + 20 = 30 (the new total is 30)

  Day 3: 30 + 5 = 35 (the final running total is 35)

  The resulting cumulative sum sequence is [10, 30, 35].

```
cumulative_sum = np.cumsum(sales_data[:, 1:], axis=1)
print(f"Cumulative Sales of restaurant: {cumulative_sum}")
```

### Sales plot

```
ptl.figure(figsize=(6, 4))
ptl.plot(np.mean(cumulative_sum, axis=0))
ptl.title("Average cumulative sales across all restaurant")
ptl.xlabel("Years")
ptl.ylabel("Sales")
ptl.grid(True)
ptl.show()
```

## Vector Operations

```
vec1 = np.array([9, 3, 1, 6, 2])
vec2 = np.array([4, 1, 5, 8, 2])
```

#### 1. Addition

```
print(f"Vector Addition: {vec1 + vec2}")
```

#### 2. Difference

```
print(f"Differences between Vector 1 and Vector 2: {vec1 - vec2}")
```

#### 3. Multiplication

```
print(f"products between Vector 1 and Vector 2: {vec1 * vec2}")
```

#### 4. Division

```
print(f"Division between Vector 1 and Vector 2: {vec1 / vec2}")
```

### Dot Product

```
dot_pro = np.dot(vec1, vec2)
print(f"Dot product: {dot_pro}")
```

### Angle



```
angle = np.arccos(np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2)))
print(angle)
```
